# 04 — Multi-Agente AML/FT

Este notebook demonstra o fluxo sequencial de cinco agentes usado na T4.


## Objetivo

Executar um workflow auditável: Dados → Detecção → Investigação → Reporte → Compliance.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT / 'src'))

from agents import run_workflow
```

```python
results = run_workflow(project_root=PROJECT_ROOT)
len(results)
```

```python
for result in results:
    print(result.agent_name)
    print('-' * 80)
    for finding in result.findings:
        print(f'- {finding}')
    print()
```

## Interpretação

O fluxo é sequencial por desenho. Cada agente recebe o contexto anterior e entrega uma saída revisável para o próximo. Em AML isso é importante porque facilita auditoria, revisão humana e explicação para compliance.
